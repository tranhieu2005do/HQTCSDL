```python
import json
```

```python
with open("../output/fixture_headtohead.json", "r", encoding="utf-8") as f:
    raw_data = json.load(f)

print(f"Loaded {len(raw_data)} fixture records.")
```

```python
head_to_head_data = []
id = 1
for raw in raw_data:
    response = raw["data"]["response"]
    for item in response:
        teams = item["teams"]
        head_to_head_data.append({
            "id" : id,
            "home_team_id" : teams["home"]["id"],
            "home_team_name" : teams["home"]["name"],
            "away_team_id" : teams["away"]["id"],
            "away_team_name" : teams["away"]["name"],
            "league_name" : item["league"]["name"],
            "home_goals" : item["goals"]["home"],
            "away_goals": item["goals"]["away"],
            "date": item["fixture"]["date"].split("T")[0]
        })
        id += 1
```

```python
print(len(head_to_head_data))
for event in head_to_head_data:
    print(event)
```

```python
with open("../database/head_to_head.json", "w", encoding="utf-8") as f:
    json.dump(head_to_head_data, f, ensure_ascii=False, indent=4)
```
