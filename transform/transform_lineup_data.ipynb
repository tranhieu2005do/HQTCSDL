```python
import json
```

```python
with open("../output/lineups.json", "r", encoding="utf-8") as f:
    raw_data = json.load(f)

print(f"Loaded {len(raw_data)} fixture records.")
```

```python
coach_data = [] # dữ liệu huấn luyện viên
lineup_data = [] # dữ liệu đội hình ra sân theo fixture_id
player_lineup_data = [] # bảng nối mối quan hệ giữa lineup_fixture và player_lineup 
```

<h2>Transform dữ liệu huấn luyện viên, sơ đồ xuất phát, danh sách xuất phát, danh sách dự bị</h2>

```python
lineup_id = 1
player_lineup_id = 1
coach_id = []
for raw in raw_data:
    data = raw["data"]["response"]
    fixture_id = int(raw["data"]["parameters"]["fixture"])
    for data_team in data:
        team = data_team["team"]

        # lọc dữ liệu huấn luyện viên
        coach = data_team["coach"]
        if coach["id"] not in coach_id:
            coach_data.append(coach)
            coach_id.append(coach["id"])
            if not coach["id"]:
                print(raw)

        # dữ liệu thông tin đội bóng như huấn luyện viện lúc đó, sơ đồ đội hình xuất phát, theo trận đấu.
        lineup_data.append({
            "id" : lineup_id,
            "fixture_id" : fixture_id,
            "team_id" : team["id"],
            "coach_id" : coach["id"],
            "formation" : data_team["formation"]
        })
        lineup_id += 1

        # danh sách cầu thủ ra sân
        player_start = data_team["startXI"]
        for player_data in player_start:
            player = player_data["player"]
            player_lineup_data.append({
                "id" : player_lineup_id,
                "fixture_id" : fixture_id,
                "team_id": team["id"],
                "player_id" : player["id"],
                "number" : player["number"],
                "position" : player["pos"],
                "grid": player["grid"],
                "status" : "Start"
            })
            player_lineup_id += 1

        # danh sách cầu thủ dự bị
        player_substitutes = data_team["substitutes"]
        for player_data in player_substitutes:
            player = player_data["player"]
            player_lineup_data.append({
                "id" : player_lineup_id,
                "fixture_id" : fixture_id,
                "team_id": team["id"],
                "player_id" : player["id"],
                "number" : player["number"],
                "position" : player["pos"],
                "grid": player["grid"],
                "status" : "Substitute"
            })
            player_lineup_id += 1
```

```python
print(len(coach_data))
for coach in coach_data:
    print(coach)
```

```python
print(len(lineup_data))
for lineup in lineup_data:
    print(lineup)
```

```python
print(len(player_lineup_data))
for player_lineup in player_lineup_data:
    print(player_lineup)
```
