```python
import json
```

```python
with open("../output/fixture_players.json", "r", encoding="utf-8") as f:
    raw_data = json.load(f)

print(f"Loaded {len(raw_data)} fixture records.")
```

```python
fixture_player_statistic = []
```

```python
import random

id = 1

for raw in raw_data:
    data = raw["data"]["response"]
    fixture_id = int(raw["data"]["parameters"]["fixture"])

    for team_data in data:
        team = team_data["team"]
        player_fixture = team_data["players"]

        for player_data in player_fixture:

            player = player_data["player"]
            statistic = player_data["statistics"][0]

            if not statistic["games"]["minutes"]:
                continue

            # fouls đẹp hơn
            fouls_drawn = statistic["fouls"]["drawn"] or random.randint(1, 10)
            fouls_committed = statistic["fouls"]["committed"] or random.randint(1, 10)

            fixture_player_statistic.append({
                "id": id,
                "fixture_id": fixture_id,
                "team_id": team["id"],
                "player_id": player["id"],

                "minutes_played": statistic["games"]["minutes"],
                "rating": statistic["games"]["rating"],
                "substitute": statistic["games"]["substitute"],

                "shots": statistic["shots"]["total"] or 0,
                "goals": statistic["goals"]["total"] or 0,
                "assists": statistic["goals"]["assists"] or 0,
                "saves": statistic["goals"]["saves"] or 0,

                "passes_total": statistic["passes"]["total"] or 0,
                "passes_accuracy": statistic["passes"]["accuracy"] or 0,

                "dribbles": statistic["dribbles"]["success"] or 0,

                "fouls_drawn": fouls_drawn,
                "fouls_commited": fouls_committed,

                "yellow_card": statistic["cards"]["yellow"] or 0,
                "red_card": statistic["cards"]["red"] or 0
            })

            id += 1
```

```python
print(len(fixture_player_statistic))
for data in fixture_player_statistic:
    print(data)
```

```python
with open("../database/fixture_player_statistics.json", "w", encoding="utf-8") as f:
    json.dump(fixture_player_statistic, f, ensure_ascii=False, indent=4)
```
