```python
import json
```

```python
with open("../output/teams.json", "r", encoding="utf-8") as f:
    raw = json.load(f)

team_info = raw[0]["data"]["response"]
print(team_info)
```

```python
team_data = []
venue_data = []
for item in team_info:
    team = item["team"]
    team_data.append({
        "id" : team["id"],
        "name" : team["name"],
        "code": team["code"],
        "founded": team["founded"],
        "logo_url": team["logo"]
    })
    venue = item["venue"]
    venue_data.append({
        "id" : venue["id"],
        "name" : venue["name"],
        "address" : venue["address"],
        "city": venue["city"],
        "capacity": venue["capacity"],
        "image_url": venue["image"]
    })
```

```python
for team in team_data:
    print(team)
```

```python
for venue in venue_data:
    print(venue)
```

<h2>Lưu data đã được transform dưới dạng json</h2>

```python
# lưu teams.json
with open("../database/teams.json", "w", encoding="utf-8") as f:
    json.dump(team_data, f, ensure_ascii=False, indent=4)

# lưu venues.json
with open("../database/venues.json", "w", encoding="utf-8") as f:
    json.dump(venue_data, f, ensure_ascii=False, indent=4)

print("Saved successfully")
```
