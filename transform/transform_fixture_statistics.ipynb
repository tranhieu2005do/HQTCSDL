```python
import json
import random
```

```python
with open("../output/fixture_statistics.json", "r", encoding="utf-8") as f:
    raw_data = json.load(f)

print(f"Loaded {len(raw_data)} fixture records.")
```

```python
fixture_statistic_data = []
```

```python
id = 1
for raw in raw_data:
    data = raw["data"]["response"]
    for team_data in data:
        team_id = team_data["team"]["id"]
        transform_data = {}
        transform_data["id"] = id
        transform_data["team_id"] = team_id
        for type in team_data["statistics"]:
            transform_data[type["type"]] = type["value"] or 0

        # xử lý null dữ liệu expected_goals
        if (transform_data.get("expected_goals") is None):
            transform_data["expected_goals"] = round(random.uniform(0, 2),2)
        fixture_statistic_data.append(transform_data)
        id += 1
```

```python
print(len(fixture_statistic_data))
for fixture in fixture_statistic_data:
    print(fixture)
```
